```python
PRICE_PER_KWH=0.10
# ensure numeric
fusion_perf["Nodes"] = pd.to_numeric(fusion_perf["Nodes"], errors="coerce")
fusion_perf["Power_per_Node_kW"] = pd.to_numeric(fusion_perf["Power_per_Node_kW"], errors="coerce")
fusion_perf["Runtime_h"] = pd.to_numeric(fusion_perf["Runtime_h"], errors="coerce")
fusion_perf["Work_Units"] = pd.to_numeric(fusion_perf["Work_Units"], errors="coerce")

# instantaneous power
fusion_perf["Power_draw_kW"] = fusion_perf["Nodes"] * fusion_perf["Power_per_Node_kW"]
fusion_perf["Power_draw_MW"] = fusion_perf["Power_draw_kW"] / 1000.0

# lines
fusion_perf["Energy_per_job_kWh"] = fusion_perf["Runtime_h"] * fusion_perf["Power_draw_kW"]
fusion_perf["Science_per_kWh"] = fusion_perf["Work_Units"] / fusion_perf["Energy_per_job_kWh"]
fusion_perf["Cost_per_job_$"] = PRICE_PER_KWH * fusion_perf["Energy_per_job_kWh"]
fusion_perf["Science_per_$"] = fusion_perf["Work_Units"] / fusion_perf["Cost_per_job_$"]

fusion_perf.head()
```

```python
nonfusion_perf["Nodes"] = pd.to_numeric(nonfusion_perf["Nodes"], errors="coerce")
nonfusion_perf["Power_per_Node_kW"] = pd.to_numeric(nonfusion_perf["Power_per_Node_kW"], errors="coerce")
nonfusion_perf["Runtime_h"] = pd.to_numeric(nonfusion_perf["Runtime_h"], errors="coerce")
nonfusion_perf["Work_Units"] = pd.to_numeric(nonfusion_perf["Work_Units"], errors="coerce")

nonfusion_perf["Power_draw_kW"] = nonfusion_perf["Nodes"] * nonfusion_perf["Power_per_Node_kW"]
nonfusion_perf["Power_draw_MW"] = nonfusion_perf["Power_draw_kW"] / 1000.0

nonfusion_perf["Energy_per_job_kWh"] = nonfusion_perf["Runtime_h"] * nonfusion_perf["Power_draw_kW"]
nonfusion_perf["Science_per_kWh"] = nonfusion_perf["Work_Units"] / nonfusion_perf["Energy_per_job_kWh"]
nonfusion_perf["Cost_per_job_$"] = PRICE_PER_KWH * nonfusion_perf["Energy_per_job_kWh"]
nonfusion_perf["Science_per_$"] = nonfusion_perf["Work_Units"] / nonfusion_perf["Cost_per_job_$"]

nonfusion_perf.head()
```

```python
# choose nodes column
if "Nodes" in fusion_arch.columns:
    fusion_arch["Nodes_eff"] = pd.to_numeric(fusion_arch["Nodes"], errors="coerce")
elif "Average Nodes" in fusion_arch.columns:
    fusion_arch["Nodes_eff"] = pd.to_numeric(fusion_arch["Average Nodes"], errors="coerce")
else:
    fusion_arch["Nodes_eff"] = pd.NA

fusion_arch["Power_per_Node_kW"] = pd.to_numeric(fusion_arch.get("Power_per_Node_kW", pd.NA), errors="coerce")
fusion_arch["Runtime_h"] = pd.to_numeric(fusion_arch.get("Runtime_h", pd.NA), errors="coerce")
fusion_arch["Work_Units"] = pd.to_numeric(fusion_arch.get("Work_Units", pd.NA), errors="coerce")

# instantaneous power
fusion_arch["Power_draw_kW"] = fusion_arch["Nodes_eff"] * fusion_arch["Power_per_Node_kW"]
fusion_arch["Power_draw_MW"] = fusion_arch["Power_draw_kW"] / 1000.0

# lines
fusion_arch["Energy_per_job_kWh"] = fusion_arch["Runtime_h"] * fusion_arch["Power_draw_kW"]
fusion_arch["Science_per_kWh"] = fusion_arch["Work_Units"] / fusion_arch["Energy_per_job_kWh"]
fusion_arch["Cost_per_job_$"] = PRICE_PER_KWH * fusion_arch["Energy_per_job_kWh"]
fusion_arch["Science_per_$"] = fusion_arch["Work_Units"] / fusion_arch["Cost_per_job_$"]

fusion_arch.head()
```

```python
if "Nodes" in nonfusion_arch.columns:
    nonfusion_arch["Nodes_eff"] = pd.to_numeric(nonfusion_arch["Nodes"], errors="coerce")
elif "Average Nodes" in nonfusion_arch.columns:
    nonfusion_arch["Nodes_eff"] = pd.to_numeric(nonfusion_arch["Average Nodes"], errors="coerce")
else:
    nonfusion_arch["Nodes_eff"] = pd.NA

nonfusion_arch["Power_per_Node_kW"] = pd.to_numeric(nonfusion_arch.get("Power_per_Node_kW", pd.NA), errors="coerce")
nonfusion_arch["Runtime_h"] = pd.to_numeric(nonfusion_arch.get("Runtime_h", pd.NA), errors="coerce")
nonfusion_arch["Work_Units"] = pd.to_numeric(nonfusion_arch.get("Work_Units", pd.NA), errors="coerce")

nonfusion_arch["Power_draw_kW"] = nonfusion_arch["Nodes_eff"] * nonfusion_arch["Power_per_Node_kW"]
nonfusion_arch["Power_draw_MW"] = nonfusion_arch["Power_draw_kW"] / 1000.0

nonfusion_arch["Energy_per_job_kWh"] = nonfusion_arch["Runtime_h"] * nonfusion_arch["Power_draw_kW"]
nonfusion_arch["Science_per_kWh"] = nonfusion_arch["Work_Units"] / nonfusion_arch["Energy_per_job_kWh"]
nonfusion_arch["Cost_per_job_$"] = PRICE_PER_KWH * nonfusion_arch["Energy_per_job_kWh"]
nonfusion_arch["Science_per_$"] = nonfusion_arch["Work_Units"] / nonfusion_arch["Cost_per_job_$"]

nonfusion_arch.head()
```

```python
# Make clean keys for fusion tables
for df in [fusion_perf, fusion_arch]:
    if "Project_Name" not in df.columns:
        df["Project_Name"] = ""
    df["Workload_key"] = df["Workload"].astype(str).str.strip().str.lower()
    df["Project_key"]  = df["Project_Name"].astype(str).str.strip().str.lower()

# Optional: drop duplicate project rows per key to avoid many to many
fusion_perf_dedup = fusion_perf.sort_values(["Workload_key","Project_key"]).drop_duplicates(["Workload_key","Project_key"], keep="first")
fusion_arch_dedup = fusion_arch.sort_values(["Workload_key","Project_key"]).drop_duplicates(["Workload_key","Project_key"], keep="first")

# Outer merge on keys only
fusion_merged = fusion_perf_dedup.merge(
    fusion_arch_dedup,
    on=["Workload_key","Project_key"],
    how="outer",
    suffixes=("_perf","_arch")
)

# Restore readable labels by coalescing the two sides
fusion_merged["Workload"] = fusion_merged.pop("Workload_perf").combine_first(fusion_merged.pop("Workload_arch"))
fusion_merged["Project_Name"] = fusion_merged.pop("Project_Name_perf").combine_first(fusion_merged.pop("Project_Name_arch"))

# Reorder columns: labels first, then everything else
fusion_cols_front = ["Workload","Project_Name","Workload_key","Project_key"]
fusion_merged = fusion_merged[fusion_cols_front + [c for c in fusion_merged.columns if c not in fusion_cols_front]]

fusion_merged.shape, fusion_merged.head()
```

```python

# Make clean keys for nonfusion tables
for df in [nonfusion_perf, nonfusion_arch]:
    if "Project_Name" not in df.columns:
        df["Project_Name"] = ""
    df["Workload_key"] = df["Workload"].astype(str).str.strip().str.lower()
    df["Project_key"]  = df["Project_Name"].astype(str).str.strip().str.lower()

# Optional: drop duplicate project rows per key
nonfusion_perf_dedup = nonfusion_perf.sort_values(["Workload_key","Project_key"]).drop_duplicates(["Workload_key","Project_key"], keep="first")
nonfusion_arch_dedup = nonfusion_arch.sort_values(["Workload_key","Project_key"]).drop_duplicates(["Workload_key","Project_key"], keep="first")

# Outer merge on keys only
nonfusion_merged = nonfusion_perf_dedup.merge(
    nonfusion_arch_dedup,
    on=["Workload_key","Project_key"],
    how="outer",
    suffixes=("_perf","_arch")
)

# Restore readable labels
nonfusion_merged["Workload"] = nonfusion_merged.pop("Workload_perf").combine_first(nonfusion_merged.pop("Workload_arch"))
nonfusion_merged["Project_Name"] = nonfusion_merged.pop("Project_Name_perf").combine_first(nonfusion_merged.pop("Project_Name_arch"))

# Reorder columns
nonfusion_cols_front = ["Workload","Project_Name","Workload_key","Project_key"]
nonfusion_merged = nonfusion_merged[nonfusion_cols_front + [c for c in nonfusion_merged.columns if c not in nonfusion_cols_front]]

nonfusion_merged.shape, nonfusion_merged.head()
```
